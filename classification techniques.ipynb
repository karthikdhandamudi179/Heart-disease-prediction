```python
df['target']=(df['target']>0).astype(int)
print(f"Healthy:{sum(df['target']==0)}")
print(f"Heart Disease:{sum(df['target']==1)}")
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

# Decision tree

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(max_depth=5,random_state=42)
dt_model
dt_model.fit(x_train,y_train)
dt_pred=dt_model.predict(x_test)
from sklearn.metrics import accuracy_score
dt_acc=accuracy_score(y_test,dt_pred)
dt_acc
```

# RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
rf_model=RandomForestClassifier(max_depth=5,random_state=42)
rf_model
rf_model.fit(x_train,y_train)
rf_pred=rf_model.predict(x_test)
from sklearn.metrics import accuracy_score
rf_acc=accuracy_score(y_test,rf_pred)
print(rf_acc)
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5,
                               metric='euclidean',
                               weights='uniform')
knn_model
knn_model.fit(x_train,y_train)
knn_pred=knn_model.predict(x_test)
from sklearn.metrics import accuracy_score
knn_acc=accuracy_score(y_test,knn_pred)
print(knn_acc)
knn_model
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lr_model=LogisticRegression(max_iter=1000)
lr_model
lr_model.fit(x_train,y_train)
lr_pred=lr_model.predict(x_test)
from sklearn.metrics import accuracy_score
lr_acc=accuracy_score(y_test,lr_pred)
print(lr_acc)
```

note : max_iter=1000, random_state=42 only use when accuracy is less

scaled data only use that time when accuracy is very low for Lgr

# Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
nb_model=GaussianNB()
nb_model
nb_model.fit(x_train,y_train)
nb_pred=nb_model.predict(x_test)
from sklearn.metrics import accuracy_score
nb_acc=accuracy_score(y_test,nb_pred)
print(nb_acc)
```

note: nb scaling is not required
Note: Naive Bayes is probabilty based model since scaled is not required 
but in worest condition we used

# SVM.SVClassifier

```python
from sklearn.svm import SVC
svc_model=SVC(kernel='rbf', C=1)
svc_model
svc_model.fit(x_train_scaled,y_train)
svc_pred=svc_model.predict(x_test_scaled)
from sklearn.metrics import accuracy_score
svc_acc=accuracy_score(y_test,svc_pred)
print(svc_acc)
svc_model
```

Notes IN svn its Distance based since scaling is manadatory.

# Model camparision

```python
model_accuracies={
    'DT':dt_acc,
    'RF':rf_acc,
    'SVM':svc_acc,
    'LgR':lr_acc,
    'NB':nb_acc,
    'KNN':knn_acc
}
model_accuracies
```

```python
best_model_name=max(model_accuracies, key=model_accuracies.get)
best_accuracy=model_accuracies[best_model_name]
print(f"\n Best Model: {best_model_name} with accuracy of {best_accuracy:4f}")
```
